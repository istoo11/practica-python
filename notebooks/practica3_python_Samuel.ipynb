# Ejercicio 05 - Funciones propias para el cálculo de importes

```python

def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    bruto = unidades * precio_unitario
    neto = bruto * (1 - (descuento_pct / 100))
    return round(neto, 2)

def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    bruto = unidades * precio_unitario
    ahorro = bruto * (descuento_pct / 100)
    neto = bruto - ahorro
    return (round(bruto, 2), round(ahorro, 2), round(neto, 2))

help(calcular_importe)

bruto, ahorro, neto = resumen_pedido(unidades=3, precio_unitario=59.90, descuento_pct=10)
print("Bruto:", bruto, "€    Ahorro:", ahorro, "€    Neto:", neto, "€")

pedidos = [(3, 59.90, 10),(1, 279.00, 0),(5, 18.50, 20),(2, 149.00, 5),(4, 89.90, 15)]

importes = []
for ped in pedidos:
    importe_calculado = calcular_importe(ped[0], ped[1], ped[2])
    importes.append(importe_calculado)

print("\nLista de importes :", importes)
print("Número de pedidos :", len(importes))
print("Importe máximo    :", max(importes), "€")
print("Importe mínimo    :", min(importes), "€")
print("Importes ordenados:", sorted(importes))
```

# Ejercicio 11 - Estadística descriptiva sobre telemetría simulada

```python
import numpy as np

rng = np.random.default_rng(2024)
temperatura = rng.normal(68, 6.5, 500)
vibracion = np.abs(rng.normal(2.4, 0.8, 500))

media_temp = np.mean(temperatura)
mediana_temp = np.median(temperatura)
std_temp = np.std(temperatura)

print(f"Media  : {media_temp:.2f}")
print(f"Mediana: {mediana_temp:.2f}")
print(f"Desv   : {std_temp:.2f} | Min: {np.min(temperatura):.2f} | Max: {np.max(temperatura):.2f}")


percentiles_vib = np.percentile(vibracion, [25, 50, 75, 95])
print("\n3. Percentiles de vibración (25, 50, 75, 95):")
print("  ", np.round(percentiles_vib, 2))

matriz_telemetria = np.column_stack((temperatura, vibracion))
print(matriz_telemetria.shape)

matriz_corr = np.corrcoef(temperatura, vibracion)
print(np.round(matriz_corr, 3))

umbral = media_temp + (2 * std_temp)
superan_umbral = np.sum(temperatura > umbral)
porcentaje = (superan_umbral / len(temperatura)) * 100

print(f"\n6. Umbral calculado (Media + 2*Std): {umbral:.2f} °C")
print(f"   Lecturas que lo superan: {superan_umbral} ({porcentaje:.1f}%)")
```

# Ejercicio 17 — if / elif / else aplicado a mantenimiento predictivo

```python
import pandas as pd
def nivel_riesgo(temperatura):
    if temperatura > 85:
        return "CRÍTICO"
    elif temperatura > 76:
        return "ALTO"
    elif temperatura > 70:
        return "MEDIO"
    else:
        return "BAJO"

print(nivel_riesgo(92.5))
print(nivel_riesgo(80.0))
print(nivel_riesgo(72.3))
print(nivel_riesgo(64.1))

sensores = pd.read_csv("../data/sensores_planta.csv")
sensores["riesgo"] = sensores["temperatura_c"].apply(nivel_riesgo)
print(sensores[["id_maquina", "temperatura_c", "riesgo"]].head())

print(sensores["riesgo"].value_counts())
filtrado = sensores[sensores["riesgo"] == "CRÍTICO"]
print(filtrado["id_maquina"].value_counts())
```
