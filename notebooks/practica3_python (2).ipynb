# Ejercicio 02 - Inventario multialmacén con listas anidadas

```python
inventario = [ ["ALM-NORTE", "Robot Aspirador", 34, 201.50],["ALM-NORTE", "Monitor 27", 58, 142.00],["ALM-SUR",   "Silla Ergonomica", 12, 178.90],["ALM-SUR",   "Robot Aspirador", 7, 201.50],["ALM-ESTE",  "Cafetera Expres", 41, 131.20],["ALM-ESTE",  "Monitor 27", 25, 142.00]]

print(inventario[3][1])

valor_stock = []

for fila in inventario:
    producto = fila[1]
    unidades = fila[2]
    precio = fila[3]
    valor = round(unidades * precio, 2)
    valor_stock.append([producto, valor])

print(valor_stock)

valor_total = 0
for fila in valor_stock:
    valor_total = valor_total + fila[1]

print("Valor total:", round(valor_total, 2))

alm_sur = inventario[2:4]
```

Si dos almacenes dejan de estar contiguos, el slicing puede incluir filas de otro almacén o dejar fuera filas que pertenecen al almacén que queremos buscar.  
Por eso el slicing depende de la posición de las filas y no es una estrategia robusta para filtrar datos.

# Ejercicio 05 - Funciones propias para el cálculo de importes

```python

def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    bruto = unidades * precio_unitario
    neto = bruto * (1 - (descuento_pct / 100))
    return round(neto, 2)

def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    bruto = unidades * precio_unitario
    ahorro = bruto * (descuento_pct / 100)
    neto = bruto - ahorro
    return (round(bruto, 2), round(ahorro, 2), round(neto, 2))

help(calcular_importe)

bruto, ahorro, neto = resumen_pedido(unidades=3, precio_unitario=59.90, descuento_pct=10)
print("Bruto:", bruto, "€    Ahorro:", ahorro, "€    Neto:", neto, "€")

pedidos = [(3, 59.90, 10),(1, 279.00, 0),(5, 18.50, 20),(2, 149.00, 5),(4, 89.90, 15)]

importes = []
for ped in pedidos:
    importe_calculado = calcular_importe(ped[0], ped[1], ped[2])
    importes.append(importe_calculado)

print("\nLista de importes :", importes)
print("Número de pedidos :", len(importes))
print("Importe máximo    :", max(importes), "€")
print("Importe mínimo    :", min(importes), "€")
print("Importes ordenados:", sorted(importes))
```
